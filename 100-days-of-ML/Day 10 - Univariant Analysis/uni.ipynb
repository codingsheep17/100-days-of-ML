```python
import matplotlib.pyplot as plt
```

```python
# 1. Count occurrences of each unique label ('male' and 'female')
counts = df['sex'].value_counts()

# 2. Define a Seaborn color palette matching the number of categories (2)
colors = sns.color_palette('pastel')[0:len(counts)]

# 3. Plot the pie chart using Matplotlib
plt.pie(
    counts, 
    labels=counts.index, 
    colors=colors, 
    autopct='%1.1f%%', 
    startangle=90
)

# 4. Add a title and display
plt.title('Gender Distribution')
plt.show()
```
