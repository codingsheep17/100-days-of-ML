```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("D:/Desktop/EDA/Data for EDA/uni.csv")
```

```python
sns.scatterplot(data=df, x=df["total_bill"], y=df["tip"], hue="time")
```

```python
sns.scatterplot(data=df, x=df["tip"], y=df["smoker"], hue="sex")
```

```python
# 1. Compute correlation for numerical columns only
corr_num = df.corr(numeric_only=True)

# 2. Plot the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_num, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Numerical Columns Correlation Heatmap")
plt.show()
```
